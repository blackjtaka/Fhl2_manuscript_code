```python
import scanpy as sc
import os
import re
from pathlib import Path

# フォルダ内のすべての .h5 ファイルを取得
data_dir = "/Users/takahiro/Desktop/project/cellbender_h5files/"
h5_files = [f for f in Path(data_dir).glob("*.h5")]

# 空のリストを用意して各ファイルを順次読み込み
adata_list = []
for h5_file in h5_files:
    # ファイルの読み込み
    adata = sc.read_10x_h5(h5_file)
    
    # サンプル名をファイル名から抽出
    sample_name = h5_file.stem.split("_")[0]  # ファイル名の最初の部分をサンプル名として使用
    adata.obs["sample"] = sample_name  # sample列に追加
    adata.var_names_make_unique()
    
    # LVRR列を作成して分類
    if re.search("V51|V30|V56|V109", sample_name):
        adata.obs["LVRR"] = "nonLVRR"
    elif re.search("V33|V64|V70|V84", sample_name):
        adata.obs["LVRR"] = "LVRR"
    elif re.search("C", sample_name):
        adata.obs["LVRR"] = "Control"
    else:
        adata.obs["LVRR"] = "Unknown"  # マッチしない場合は "Unknown" とする
    
    # 読み込んだオブジェクトをリストに追加
    adata_list.append(adata)
```

```python
# anndata.concatを使用してデータを統合
adata = ad.concat(adata_list, join="outer", label="sample", keys=[adata.obs["sample"].iloc[0] for adata in adata_list])

# 統合データの確認
print(adata)
```

```python
# 各細胞タイプに対応するクラスタ番号の辞書
cell_type_dict = {
    "Cardiomyocytes": ["1", "7"],
    "Fibroblasts": ["0"],
    "Endothelial Cells": ["2","6"],
    "Myeloid Cells": ["4"],
    "Pericytes": ["3"],
    "Lymphoid Cells": ["5"],
    "Neural Cells": ["10"],
    "Lymphatic Endothelial Cells": ["11"],
    "Smooth Muscle Cells":["9"],
    "Doublets":["8","12"]
}

# クラスタ番号と細胞タイプのマッピングを作成
leiden_to_cell_type = {cluster: cell_type for cell_type, clusters in cell_type_dict.items() for cluster in clusters}

# adata.obs["cell_type"]にマッピングを適用
adata.obs["cell_type"] = adata.obs["leiden"].map(leiden_to_cell_type)
```
