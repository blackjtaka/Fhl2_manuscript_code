```r
library("ArchR")
library(patchwork)
.libPaths()
setwd("~/workspace/reha/archr_Fhl2/")

addArchRThreads(threads = 16) 
addArchRGenome("mm10")
```

# Load data

```r
inputFiles <- c("~/workspace/reha/data/SED1/fragments.tsv.gz","~/workspace/reha/data/SED2/fragments.tsv.gz","~/workspace/reha/data/Ex1/fragments.tsv.gz","~/workspace/reha/data/Ex2/fragments.tsv.gz","~/workspace/reha/data/Fhl2OE1/fragments.tsv.gz","~/workspace/reha/data/Fhl2OE2/fragments.tsv.gz")
names(inputFiles) <- c("SED1","SED2","Ex1","Ex2","Fhl2OE1","Fhl2OE2")


# create arrow files ------------------------------------------------------
ArrowFiles <- createArrowFiles(
  inputFiles = inputFiles,
  sampleNames = names(inputFiles),
  minTSS = 3, #Dont set this too high because you can always increase later
  minFrags = 4000, 
  addTileMat = TRUE,
  addGeneScoreMat = TRUE
)
```

```r
# Compute doublet score ----------------------------------------------------

ArrowFiles <- c("SED1.arrow","SED2.arrow","Ex1.arrow","Ex2.arrow","Fhl2OE1.arrow","Fhl2OE2.arrow")
doubScores <- addDoubletScores(
  input = ArrowFiles,
  k = 10, #Refers to how many cells near a "pseudo-doublet" to count.
  knnMethod = "UMAP", #Refers to the embedding to use for nearest neighbor search with doublet projection.
  LSIMethod = 1
)
```

```r
# Create project file -----------------------------------------------------

projReha <- ArchRProject(
  ArrowFiles = ArrowFiles, 
  outputDirectory = "ArchR_result",
  copyArrows = TRUE #This is recommened so that if you modify the Arrow files you have an original copy for later usage.
)

getAvailableMatrices(projReha)
projReha <- filterDoublets(ArchRProj = projReha)
```

# Start analysis

```r
# Dimensionality Reduction and Clustering ---------------------------------

projReha <- addIterativeLSI(ArchRProj = projReha, useMatrix = "TileMatrix", name = "IterativeLSI")
```

```r
projReha <- addClusters(input = projReha, reducedDims = "IterativeLSI", resolution=0.5, force = TRUE)
```

```r
# Assigning Clusters with Gene Scores -------------------------------------

projReha <- addImputeWeights(projReha)
```

```r
markerGenes  <- c(
  "Ttn", "Mybpc3", #CMs
  "Dcn", "Col1a1", "Postn", # FBs
  "Frmd3", "Dlc1", "Myh11", # SMCs
  "Pecam1", "Cdh5", "Vwf","Npr3", # Enforthelial, cardialcells
  "Ptprc", "Mrc1", "Cd163", # Macrophages
  "Cd3e","Skap1", "CD79a", "CD79b","Il7r", "Kit", # Tcell Bcell
  "Lmnb1","Slpi","Retnlg","S100a9", #Granulocytes
  "Nrxn1", "Nrxn3", "Upk3b","Msln","Gpc3", "Fmod",
  "Plin1", "XKR4", "Acta2","Ms4a1","Ncr1","Vtn","Colec11","Steap4","Kcnj8","Myh11","Mmrn1","Flt4"
)

p <- plotEmbedding(
  ArchRProj = projReha, 
  colorBy = "GeneScoreMatrix", 
  name = markerGenes, 
  embedding = "UMAP",
  imputeWeights = getImputeWeights(projReha),
  baseSize = 12
)
```
