```python
import pertpy as pt
import scanpy as sc
```

```python
adata_full = sc.read_h5ad("./adata/rawadata_curated_Ctrl_6W.h5ad")
```

```python
adata_full
```

```python
adata = adata_full[adata_full.obs["type"].isin(["Reha6W","SED6W"])].copy()
```

```python
adata
```

```python
ag_rfc = pt.tl.Augur("random_forest_classifier")
```

```python
loaded_data = ag_rfc.load(adata, label_col="type", cell_type_col="cell_type")
```

## subsample_size 50

```python
v_adata, v_results = ag_rfc.predict(
    loaded_data, subsample_size=50, n_threads=6, select_variance_features=False, span=0.75, random_state=1
)
```

```python
v_results["summary_metrics"]
```

```python
lollipop = ag_rfc.plot_lollipop(v_results)
```
