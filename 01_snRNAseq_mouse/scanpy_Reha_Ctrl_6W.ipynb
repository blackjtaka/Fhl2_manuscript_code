```python
# 変数名をユニークにする
for adata in [adata1, adata2, adata3, adata4, adata5, adata6]:
    adata.var_names_make_unique()
```

```python
# サンプル名リスト
sample_names = ["Ctrl1", "Ctrl2", "Reha6W1", "Reha6W2","SED6W1", "SED6W2"]

# 変数名をユニークにし、サンプル名を追加する
adata_list = [adata1, adata2, adata3, adata4, adata5, adata6]
for adata, sample_name in zip(adata_list, sample_names):
    adata.var_names_make_unique()
    adata.obs["sample"] = sample_name

# AnnData オブジェクトを結合
adata = sc.concat(adata_list)

# 結果を確認
print(adata)
```

```python
# 各細胞タイプに対応するクラスタ番号の辞書
cell_type_dict = {
    "Cardiomyocytes": ["2", "3", "13"],
    "Fibroblasts": ["1"],
    "Endothelial Cells": ["0", "7", "8"],
    "Myeloid Cells": ["4"],
    "Pericytes": ["5"],
    "Endocardial Cells": ["6"],
    "Lympatic Endothelial Cells": ["9"],
    "Epicardial Cells": ["10"],
    "B cells": ["11"],
    "Smooth Muscle Cells": ["12"],
    "Lymphoid Cells": ["14"]
}

# クラスタ番号と細胞タイプのマッピングを作成
leiden_to_cell_type = {cluster: cell_type for cell_type, clusters in cell_type_dict.items() for cluster in clusters}

# adata.obs["cell_type"]にマッピングを適用
adata.obs["cell_type"] = adata.obs["leiden"].map(leiden_to_cell_type)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 'type'と'leiden'ごとの集計を行い、割合を計算
type_leiden_counts = adata.obs.groupby(['type', 'cell_type']).size().unstack(fill_value=0)
type_leiden_percentage = type_leiden_counts.div(type_leiden_counts.sum(axis=1), axis=0) * 100

# 積み上げ棒グラフの作成
type_leiden_percentage.plot(kind='bar', stacked=True, figsize=(10, 6))

# グラフのタイトルとラベルを設定
plt.title('Cell type Proportions per sample')
plt.xlabel('Sample')
plt.ylabel('Percentage')

# 凡例を調整
plt.legend(title='Cell type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(False)
# グラフを表示
plt.tight_layout()
plt.show()
```
